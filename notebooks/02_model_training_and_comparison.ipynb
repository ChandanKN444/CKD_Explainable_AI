```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys

root = Path("..").resolve()
if str(root) not in sys.path:
    sys.path.append(str(root))

from src.train import train_and_evaluate_all_models
from src.config import METRICS_DIR, FIGURES_DIR

%matplotlib inline
```

## 1. Execute Training and Cross-Validation Pipeline

```python
benchmarks, best_model = train_and_evaluate_all_models()
print(f"Best Performing Model: {best_model}")
```

## 2. Comparative Evaluation Metrics

```python
metrics_table = []
for name, m in benchmarks.items():
    metrics_table.append({
        "Model": name,
        "Accuracy": f"{m['accuracy']*100:.2f}%",
        "Precision": f"{m['precision']:.4f}",
        "Recall": f"{m['recall']:.4f}",
        "Specificity": f"{m['specificity']:.4f}",
        "F1-Score": f"{m['f1_score']:.4f}",
        "ROC-AUC": f"{m['roc_auc']:.4f}" if m['roc_auc'] else "N/A"
    })
pd.DataFrame(metrics_table)
```
