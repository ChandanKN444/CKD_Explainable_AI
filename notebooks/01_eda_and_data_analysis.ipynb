# Chronic Kidney Disease (CKD) - Exploratory Data Analysis
**Course Project:** CKD Prediction using Explainable AI (XAI)

This notebook conducts an in-depth clinical analysis of the UCI Chronic Kidney Disease dataset:
- Attribute statistics & missing data mapping
- Biomarker distribution analysis (Serum Creatinine, Blood Urea, Hemoglobin, Specific Gravity)
- Pearson correlation matrix & renal function interactions

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys

# Set project root
root = Path("..").resolve()
if str(root) not in sys.path:
    sys.path.append(str(root))

from src.data_loader import load_raw_data, get_feature_and_target_matrices
from src.config import NUMERICAL_FEATURES, CATEGORICAL_FEATURES, FEATURE_METADATA

%matplotlib inline
plt.style.use("seaborn-v0_8-whitegrid")
```

## 1. Load Cleaned Dataset

```python
X, y = get_feature_and_target_matrices()
df = pd.concat([X, y], axis=1)
print(f"Dataset shape: {df.shape}")
df.head()
```

## 2. Target Variable Distribution

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(x="class", data=df, palette=["#38A169", "#E53E3E"])
plt.title("CKD Target Class Distribution (0 = NotCKD, 1 = CKD)", fontsize=12, fontweight="bold")
plt.xlabel("Disease Class")
plt.ylabel("Patient Count")
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())} ({p.get_height()/len(df):.1%})',
                (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                ha='center', va='center', color='white', fontweight='bold')
plt.show()
```

## 3. Key Renal Biomarker Boxplots

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
biomarkers = [("sc", "Serum Creatinine (mg/dL)"),
              ("hemo", "Hemoglobin (g/dL)"),
              ("bu", "Blood Urea (mg/dL)"),
              ("sg", "Specific Gravity")]

for idx, (col, label) in enumerate(biomarkers):
    ax = axes[idx // 2, idx % 2]
    sns.boxplot(x="class", y=col, data=df, ax=ax, palette=["#38A169", "#E53E3E"])
    ax.set_title(f"{label} by Class", fontweight="bold")
    ax.set_xticklabels(["Healthy (NotCKD)", "CKD Positive"])

plt.tight_layout()
plt.show()
```

## 4. Correlation Heatmap

```python
plt.figure(figsize=(12, 10))
corr = df[NUMERICAL_FEATURES].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, cbar=True)
plt.title("Correlation Matrix of Continuous Biomarkers", fontsize=14, fontweight="bold")
plt.show()
```
